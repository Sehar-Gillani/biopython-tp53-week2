```python
print("GC %:", gc_fraction(dna) * 100)
print("Complement:", dna.complement())
print("Reverse complement:", dna.reverse_complement())
print("RNA:", dna.transcribe())
print("Protein:", dna.translate())
```

```python
print(dna.translate(to_stop=True))
```

```python
from Bio import Entrez, SeqIO

Entrez.email = "Sehrm.gillani@gmail.com"

handle = Entrez.efetch(db="nucleotide", id="NM_000546", rettype="gb", retmode="text")
record = SeqIO.read(handle, "genbank")
handle.close()

print(record.id)
print(record.description)
print(len(record.seq))
print("GC %:", gc_fraction(record.seq) * 100)
```

```python
for feature in record.features:
    if feature.type == "CDS":
        cds = feature.extract(record.seq)
        print("CDS length:", len(cds))
        protein = cds.translate(to_stop=True)
        print("Protein length:", len(protein))
        print(protein[:50])
```

```python
search = Entrez.esearch(db="nucleotide", term="TP53[Gene Name] AND Homo sapiens[Organism]", retmax=5)
results = Entrez.read(search)
search.close()
print("Total found:", results["Count"])
print("First IDs:", results["IdList"])

SeqIO.write(record, "tp53.fasta", "fasta")
print("Saved tp53.fasta")
```
